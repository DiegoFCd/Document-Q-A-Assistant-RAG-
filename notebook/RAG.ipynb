## Y ahora viene el primer concepto verdaderamente importante y característico de RAG:

### ✂️ Text Splitter

### ¿Por qué es necesario dividir el documento?

Imagina que tienes un PDF de 100 páginas.
Sería muy ineficiente darle un archivo completo a nuestra app, porque consumiria demasiados tokens, habria falta de precisión debido al volumen de datos que le estamos pasando y perderiamos la granulidad de los datos.
La idea es tomar el archivo completo y dividirlo en Chunks:

✔ Chunk 1 : Ej. Supervised learning uses labeled data...

✔ Chunk 2 : Ej. Clustering is an unsupervised technique...

✔ Chunk 3 : Ej. Random Forest is an ensemble method...


### ¿Qué Splitter utilizaré? 

*RecursiveCharacterTextSplitter*

Este es uno de los más utilizados

## 1- Imports

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

## 2- Crear el Splitter

```python
text_splitter = RecursiveCharacterTextSplitter (
    chunk_size = 1000,
    chunk_overlap = 200
)
# Cada fragmento tendrá aproximadamente: 1000 caracteres
# Los chunks compartirán: 200 caracteres entre si para no perder contexto. 
# Ej: Chunk 1: ABCDEFGHIJ   /    Chunk 2: HIJKLMNO 
```

## 3- Aplicar esa configuración al documento

```python
chunks = text_splitter.split_documents(documents)
```

## 4- Chequear cantidad de chunks se generaron

```python
print(len(chunks))
```

## 5- Inspeccionar el primer chunk

```python
print(chunks[0].page_content) #Agregando esta funcion evito mostrar la metaData 
```

## Y ahora sigue uno de los conceptos más importantes de todo RAG: 

## 🧠 Embeddings

 ### ¿Qué es un embedding?

Es una representación vectorial del significado del texto.

*"The cat is sleeping"*
↓
[0.12, -0.84, 0.33, ...]

*"The kitten is resting"*
↓
[0.10, -0.81, 0.30, ...]  Cada vector representa una palabra

### La magia detrás de esto?

No buscamos palabras. Buscamos significado.
* Por eso una pregunta: *What is clustering?*
* Puede encontrar un chunk que diga: *Clustering is an unsupervised learning technique.*
* aunque no coincidan exactamente todas las palabras

## Ahora viene 🗃️ Vector Store, Aquí es donde aparece la "memoria semántica" de nuestro asistente.

### Conceptualmente hasta ahora tenemos:

* Chunk

* [0.005, 0.002, ...]

*pero esos vectores están "sueltos".*

### Necesitamos almacenarlos en una estructura que permita:

## 2- Importo Chroma

```python
from langchain_chroma import Chroma
```

## Creo base vectorial

```python
vector_store = Chroma.from_documents(
    documents = chunks,
    embedding = embeddings
)
# Lo que ocurre aqui es que cada Chunk queda Enbebido luego es vectorizado y guardado en memoria: Algo así:

# Chunk 1
#   ↓
# Embedding
#   ↓
# Vector
#   ↓
# Guardar
```

### Conceptualmente Chroma construye algo parecido a esta tabla:

* Chunk -------------------------------------          Embedding
* "Supervised learning uses labeled data..."	[0.005,0.004,...]
* "Clustering is an unsupervised technique..."	[0.014,-0.021,...]
* "Random Forest is an ensemble method..."	    [-0.008,0.013,...]
Y luego lo guarda.

### Y esto es justamente una Vector Store. Es una base de datos especializada para guardar:

- Texto -->  Representación matemática del significado

# 4- Compruebo cuántos chunks guardamos

```python
vector_store._collection.count()
```

## Y ahora sigue trabajar con 🔍 Similarity Search

### 1- Hago una busquda simple 

```python
#Formulo la pregunta
query = "Hay una superinteligencia"
```

```python
# Ejecuto busqueda
results = vector_store.similarity_search(query = query, k=3)
```

### ¿Qué ocurre internamente?
- Pregunta
- ↓
- Embedding
- ↓
- Comparar contra todos los vectores
- ↓
- Encontrar los más cercanos
- ↓
- Devolver los chunks correspondientes

### Ahora comenzamos con la parte donde se integra el uso de Gemini

### En los 'page_content' anteriores ya tenemos los resultados. Ahora se necesita unir los textos para darle contexto a Gemini cuando analize los resultados

## 1- Unimos los resultados

```python
context = "\n\n".join(
    doc.page_content
    for doc in results
)
```

## 2- Construir el Promp

```python
prompt = f"""
Answer the question based only on the context below, justifie the response.

Context:
{context}

Question:
{query}
"""
```

### Conceptualmente estamos haciendo:

- Chunks recuperados
- ↓
- Context Injection
- ↓
- Prompt

## 3- Llamar a Gemini

```python
#Imports
from langchain_google_genai import ChatGoogleGenerativeAI
```

```python
llm = ChatGoogleGenerativeAI (
    model = "gemini-2.5-flash"
)
```

```python
#Generar respuesta
response = llm.invoke(prompt)
```

```python
#Chequear respuesta
print(response.content)
```

### Y aquí aparece la verdadera magia

* Gemini NO está respondiendo con lo que aprendió durante su entrenamiento.

* Está respondiendo utilizando:

**EL PDF**

### Esto es exactamente la esencia de RAG.
